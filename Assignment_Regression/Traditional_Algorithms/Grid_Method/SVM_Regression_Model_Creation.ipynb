Importing libraries:

```python
import pandas as pd
import numpy as np
```

Load Dataset:

```python
dataset=pd.read_csv("insurance_dataset.csv")
```

Convert Nominal to Numerical datas:

```python
dataset=pd.get_dummies(dataset,drop_first=True,dtype=int)
```

Display columns:

```python
dataset.columns
```

Input/Output split:

```python
independent=dataset[['age', 'bmi', 'children','sex_male', 'smoker_yes']]
```

```python
dependent=dataset[['charges']]
```

Train/Test Split data:

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(independent,dependent,test_size=0.20,random_state=42)
```

Train a Model with help of GridSearchCV:

```python
from sklearn.model_selection import GridSearchCV
from sklearn.svm import SVR
param_grid = {'kernel':['linear', 'poly', 'rbf', 'sigmoid'],
                'C':[1000,2000,3000],
                 'gamma':['scale', 'auto']
             }

grid=GridSearchCV(SVR(),param_grid,refit=True,verbose=3,n_jobs=-1)

grid.fit(X_train, np.asarray(y_train).ravel())
```

Store the Grid results (Parameters) as Dictionary:

```python
re=grid.cv_results_
```

Display the Grid results (Parameters) as Table Format:

```python
table=pd.DataFrame(re)
table
```

Display the Best Parameters:

```python
print("The Best Parameters are ",grid.best_params_)
```

Display the CV's Avg Best  R²_Score:

```python
print("CV R² score:", grid.best_score_)
```

Model Evaluation:

```python
y_pred=grid.predict(X_test)
```

```python
from sklearn.metrics import r2_score
r_score=r2_score(y_test,y_pred)
print(f"The model performance is {r_score}")
```
